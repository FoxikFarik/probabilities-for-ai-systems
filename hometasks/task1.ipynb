**Задача 1.16** За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
# 1.16 а
2 * sc.factorial(5) * sc.factorial(5) / sc.factorial(10)
```

```python
# 1.16 б
sc.factorial(5) * 2 ** 5 / sc.factorial(10)
```

**Задача 1.16а** В аудитории m = 25 студентов. Найти вероятность того, что хотя бы у двух студентов дни рождения совпадают. При каком числе m студентов вероятность того же события не меньше чем 0,95? (Полагаем равновозможность рождений в любой день года.)

```python
# 1.16 а 1
1 - (sc.factorial(365,exact=True) / sc.factorial(365 - 25,exact=True)) / 365 ** 25
```

```python
# 1.16 а 2
p = 0
m = 25
while p < 0.95:
    m += 1
    p = 1 - (sc.factorial(365, exact=True) / sc.factorial(365 - m, exact=True)) / 365 ** m
print(m, p)
    
```
